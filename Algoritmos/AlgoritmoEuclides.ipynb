```python
def calcularFactorial(factorial):
	"""	Esta función calcula el factorial de un número entero.
	"""
	
	fac = 1
	fact = ""
	for i in range(factorial, 0, -1):
		fac *= i
		fact += imprimirSerie(i)
		
	return fac, fact
```

```python
def imprimirSerie(i):
	"""	Esta función imprime la serie números involucrados
		en el cálculo de factorial de i de calcularFactorial()
	"""

	fact = ""
	if(i == 1):
		fact += str(i)	
	else:
		fact += str(i) + "*"
	
	return fact
```

```python
def principal():
	print("\nCALCULA EL FACTORIAL\nDE UN NÚMERO ENTERO")
	factorial = 3
    #factorial = leerDato()
	fac, fact = calcularFactorial(factorial)
	print(fact + " = " + str(fac))
	return

principal()
```

```python
"""	Title: Sucesión de ULAM
	Author: Eric Torres
	UEA: Introducción a la programación
	Professor: Dra. Lizbeth Gallardo López
	Description: La sucesión de ULAM consiste en

	a) Inicia con cualquier entero positivo (dado por el usuario).

	b) Si el número es par, divídelo entre 2. Si es impar, 
	multiplícalo por 3 y agrégale 1 

	c) Repetir el paso b hasta que se obtenga un valor 1.

	El programa deberá imprimir en la pantalla la sucesión 
	de ULAM.
"""
#entero = int(input("\n Dame un número entero positivo\n>> "))
entero = 4

m = ""
while entero > 1:
	
	if entero % 2 == 0:
		entero = entero / 2
	else:
		entero = entero * 3 + 1
	
	m += str(int(entero))+" "


print("\n", m, "\n")
```
