# Algoritmo de recursividad

Recursión significada que una función se llama a sí misma.

## La pila de llamadas con recursión

```python
def fact(x):
    '''Función recursiva para calcular el factorial de un número.'''
    print(x)
    if x == 1:
        return 1
    else:
        return x * fact(x-1)
fact(3)
```

## Suma con recursión

```python
def suma(lista):
  if len(lista) == 1:   
      # Caso base
      return lista[0]
  else:      
      # La función se llama a si misma (Recursión)
      # lista[1:] devuelve la lista sin su primer elemento
      return lista[0] + suma(lista[1:]) 
                                          
suma([1,2,3,4,5])
```

```python
def sumar(array):
    '''Functión recursiva para calcular la suma de todos los números
        de un array y devolver el total.'''
    total = 0
    for x in array:
        print(total, '+', x)
        total += x
    return total
    
sumar([1,2,3,4])
```

```python
def sumarArray(array):
    if len(array) < 2:
        return array[0]
    else:
        return (array[-1]) + sumarArray(array[:-1])

sumarArray([1,2,3,4])
```
