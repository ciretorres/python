# Algoritmo Búsqueda binaria

Este es un problema de búsqueda. En todos los casos se utiliza el mismo algoritmo: *búsqueda binaria*.

```python
def busqueda_binaria(lista, elemento):
    ''' Recibe un array ordenado y un elemento.
        Si el elemento se encuentra en el array,
        la función devuelve su posición.
    '''
    menor = 0
    mayor = len(lista) - 1

    while menor <= mayor:
        medio = int((menor + mayor)/2)
        estimado = lista[medio]
        if estimado == elemento:
            return medio
        if estimado > elemento:
            mayor = medio -1 
            
        else:
            menor = medio + 1
    return 'null'            
```

```python
mi_lista = [1, 3, 5, 7, 9]
```

```python
# => 1
busqueda_binaria(mi_lista, 3)
```

```python
# => None
busqueda_binaria(mi_lista, -1)
```
