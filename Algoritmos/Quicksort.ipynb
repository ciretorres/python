# Quicksort

Los arrays vacío o con un solo elemento definen el caso base. Puedes
devolver esos array como si no hubiese nada que ordenar:

```python
def quicksort(array):
    if len(array) < 2:
        return array
    else:
        pivote = array[0]
        menores = [i for i in array[1:] if i <= pivote]

        mayores = [i for i in array[1:] if i > pivote]

        return quicksort(menores) + [pivote] + quicksort(mayores)
```

```python
quicksort([15, 10]) + [33] + quicksort( [] )
```

```python
quicksort([10, 5, 2, 3])
```
