# Tablas Hash

## Función Hash

En python con llamados diccionarios. Puedes crear una nueva tabla hash usando la función `$\dict()$`:

```python
libro = dict()
```

`libro`es una nueva tabla hash. Añadamos precios al `libro`:

```python
libro["manzana"] = 0.67
libro["leche"] = 1.49
libro["aguacate"] = 1.49
libro
```

¡Bien fácil! Ahora preguntemos el precio de un aguacate:

```python
libro["aguacate"]
```
