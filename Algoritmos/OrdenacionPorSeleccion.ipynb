```python
# Algoritmo Ordenación por selección
```

```python
def encuentraMenor(array):
    """Regresa el índice del menor elemento de un array"""
    menor = array[0] # indica el menor valor
    indice_menor = 0 # indica el índice del menor valor
    for i in range(1, len(array)):
        if array[i] < menor:
            menor = array[i]
            indice_menor = i
    return indice_menor
```

```python
def ordenacionPorSeleccion(array):
    """Regresa un array ordenado de menor a mayor por selección de índice"""
    nuevoArr = []
    for i in range(len(array)):
        # Encuentra el el menor elemento en el array y añádelo al nuevo array
        menor = encuentraMenor(array)
        print('índice', menor, '=', array[menor])
        nuevoArr.append(array.pop(menor))
    return nuevoArr
```

```python
ordenacionPorSeleccion([5, 3, 6, 2, 10])
```
